**Other typos / inconsistencies found:**

- **Genres are listed alphabetically, not by importance.** Every film's `genres` list is in alphabetical order, so the first genre is *not* the film's "main" genre (e.g. *The Lord of the Rings* is listed as Adventure, Drama, Fantasy simply because A comes before D before F).
```python
# Clean up: drop useless columns, simplify column names, fill/standardize values
# 1. Drop columns that are empty, constant, or just web links
drop_cols = ['parentImdbId', 'endYear', 'type', 'isAdult',       # empty or constant
             'url', 'primaryImage', 'thumbnails', 'trailer',     # links
             'externalLinks', 'interests']                       # links / tag lists
df_clean = df_clean.drop(columns=drop_cols)

# 2. Rename camelCase columns to snake_case
df_clean = df_clean.rename(columns={
    'primaryTitle': 'primary_title', 'originalTitle': 'original_title',
    'contentRating': 'content_rating', 'startYear': 'start_year',
    'releaseDate': 'release_date', 'countriesOfOrigin': 'countries_of_origin',
    'spokenLanguages': 'spoken_languages', 'filmingLocations': 'filming_locations',
    'productionCompanies': 'production_companies', 'grossWorldwide': 'gross_worldwide',
    'runtimeMinutes': 'runtime_minutes', 'averageRating': 'average_rating',
    'numVotes': 'num_votes'})

# 3. Standardize content-rating labels and fill the 6 missing ones
df_clean['content_rating'] = (df_clean['content_rating'].astype('string')
                              .str.strip()
                              .replace({'Unrated': 'Not Rated'})
                              .fillna('Unknown')
                              .astype('category'))

# 4. Trim title text just in case
df_clean['primary_title'] = df_clean['primary_title'].str.strip()

df_clean.head()
```

### Additional new columns

```python
# New column 1: decade of release
df_clean['decade'] = (df_clean['start_year'] // 10) * 10

# New column 2: number of genres listed for the film
df_clean['num_genres'] = df_clean['genres'].str.len()

# New column 3: gap between critics (Metascore / 10) and IMDb users
# positive = critics liked it more than IMDb users; negative = users liked it more
df_clean['critic_audience_gap'] = (df_clean['metascore'].astype('float') / 10
                                   - df_clean['average_rating']).round(2)

# New column 4: first-listed country, and a "trustworthy" USD budget
# (budgets of many non-US films are stored in local currency, e.g. yen or rupees)
df_clean['primary_country'] = df_clean['countries_of_origin'].str[0]
df_clean['budget_usd'] = df_clean['budget'].where(df_clean['primary_country'] == 'US')

# New column 5: return on budget (only where the budget is trustworthy)
df_clean['gross_to_budget'] = (df_clean['gross_worldwide'].astype('float')
                               / df_clean['budget_usd'].astype('float')).round(2)

# New column 6: runtime category
df_clean['runtime_category'] = pd.cut(df_clean['runtime_minutes'],
                                      bins=[0, 100, 150, 1000],
                                      labels=['short (<=100 min)', 'medium (101-150 min)', 'long (>150 min)'])

df_clean[['primary_title', 'start_year', 'decade', 'genres', 'num_genres',
          'metascore', 'average_rating', 'critic_audience_gap',
          'primary_country', 'budget_usd', 'gross_to_budget', 'runtime_category']].head(10)
```

```python
# How many films have a usable USD budget, and which films earned the most per dollar?
print("Films with a trustworthy USD budget:", df_clean['budget_usd'].notna().sum(), "of", len(df_clean))
df_clean.nlargest(5, 'gross_to_budget')[['primary_title', 'start_year', 'budget_usd', 'gross_worldwide', 'gross_to_budget']]
```

**What the new columns represent:**
- **`decade`** - the decade the film was made (e.g. 1990), based on `start_year`. This makes it easy to compare time periods.
- **`num_genres`** - how many genres IMDb lists for the film (1 to 3).
- **`critic_audience_gap`** - the Metascore divided by 10 minus the IMDb rating, so both are on a 0-10 scale. A **positive** number means critics liked the film more than IMDb users; a **negative** number means users liked it more than critics. It is missing when the film has no Metascore.
- **`primary_country`** - the first country listed for the film.
- **`budget_usd`** - the budget, but only for films whose first listed country is the US. Because many non-US budgets are in local currency (see section 4), we only keep the budgets we can trust to be in dollars. Only 142 of the 250 films have one.
- **`gross_to_budget`** - worldwide gross divided by the USD budget, which shows how many times a film earned back its cost. The biggest earners are classic low-budget hits like *Rocky* (about 122 times its budget) and *Jaws* (about 70 times).
- **`runtime_category`** - short (100 minutes or less), medium (101-150 minutes) or long (over 150 minutes).

```python
# Group by decade: number of films, average rating, votes, runtime
decade_summary = (df_clean.groupby('decade')
                  .agg(num_films=('primary_title', 'count'),
                       avg_rating=('average_rating', 'mean'),
                       avg_votes=('num_votes', 'mean'),
                       avg_runtime=('runtime_minutes', 'mean'),
                       avg_metascore=('metascore', 'mean'))
                  .round(2))
decade_summary
```

**What we learned (decade):** The dataset is weighted toward recent decades: the 1990s-2010s together contain 131 of the 250 films, while the 1920s and 1930s only have 6 each. The average rating is quite similar across decades (between about 8.1 and 8.4), so no decade is dramatically better, although the 1990s are the highest (8.41) and the 1920s the lowest (8.13). Two things change a lot over time: the **number of votes** (films from the 1920s-1960s average roughly 120,000-340,000 votes, while 1990s-2010s films average roughly 0.9-1.2 million, which makes sense because IMDb users are mostly younger and watch newer films) and the **Metascore** (it falls from about 92-96 for films from the 1920s-1950s to about 78-81 for films since 1980, so critics rate classics very highly).

```python
# Group by genre (a film with several genres counts once in each of its genres)
genre_summary = (df_clean.explode('genres').groupby('genres')
                 .agg(num_films=('primary_title', 'count'),
                      avg_rating=('average_rating', 'mean'),
                      avg_gross=('gross_worldwide', 'mean'),
                      avg_runtime=('runtime_minutes', 'mean'),
                      best_rating=('average_rating', 'max'))
                 .round(2)
                 .sort_values('avg_rating', ascending=False))
genre_summary
```

**What we learned (genre):** Average ratings are again very close, between about 8.1 (Sport) and 8.4 (Fantasy), so genre does not strongly decide how well a film is rated. Among the larger genres, **Fantasy** (8.41, 18 films), **Crime** (8.36, 54 films) and **Sci-Fi** (8.36, 23 films) have the highest average ratings, while **Comedy** (8.22) is the lowest of the large genres. Box office tells a different story: **Fantasy** (about $630 million), **Adventure** (about $547 million), **Sci-Fi** (about $528 million), **Action** (about $465 million) and **Animation** (about $452 million) earn far more per film than **Drama** (about $150 million), even though Drama is the biggest genre by far. So the films that make the most money are not necessarily the best rated, and Animation and Comedy films are the shortest on average (about 107-109 minutes) while History and Biography are the longest (about 146-153 minutes).
